```python
soup = BeautifulSoup(r.text, 'html.parser')
results = soup.find_all(['h1', 'p'])
text = [result.text for result in results]
arti = ' '.join(text)
```

```python
mc =100
sentences = arti.split(' ')
cc = 0 
c = []
for sentence in sentences:
    if len(c) == cc + 1: 
        if len(c[cc]) + len(sentence.split(' ')) <= mc:
            c[cc].extend(sentence.split(' '))
        else:
            cc += 1
            c.append(sentence.split(' '))
    else:
        print(cc)
        c.append(sentence.split(' '))

for chunk_id in range(len(c)):
    c[chunk_id] = ' '.join(c[chunk_id])
```

```python
res = clf(c, max_length=120, min_length=30, do_sample=False)
```
